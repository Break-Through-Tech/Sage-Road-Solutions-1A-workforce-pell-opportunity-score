```python
ipeds = pd.read_csv("data/processed/IPEDS/IPEDS_dataset_cleaned.csv")
cs_pseo = pd.read_csv("data/merges/CS_PSEO_merged.csv")
```

```python
merged = pd.merge(cs_pseo, ipeds, left_on="UNITID", right_on="unitid", how="left")
```

```python
merged.head()
```

```python
merged.columns.tolist()
```

```python
merged_cleaned = merged.drop(columns=['unitid', 'city', 'institution_name', 'STABBR', 
    'EARN_MDN_HI_1YR', 'EARN_COUNT_WNE_1YR', 'EARN_MDN_1YR', 'EARN_COUNT_WNE_5YR', 'EARN_MDN_5YR', 'ADM_RATE'])

state_col = merged_cleaned.pop('state')
city_col = merged_cleaned.pop('CITY')
bea_col = merged_cleaned.pop('bea_region')
zip_col = merged_cleaned.pop('ZIP')

merged_cleaned.insert(3, 'state', state_col)
merged_cleaned.insert(4, 'city', city_col)
merged_cleaned.insert(5, 'bea_region', bea_col)
merged_cleaned.insert(6, 'ZIP', zip_col)
```

```python
merged_cleaned.columns.tolist()
```

```python
merged_cleaned.to_csv("CS_PSEO_IPEDS_merged.csv", index=False)
```
