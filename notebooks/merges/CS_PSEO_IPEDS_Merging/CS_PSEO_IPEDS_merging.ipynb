"The 132 missing rows represent fully operational, active programs (CURROPER = 1). CA's recommendation is correct regarding new fields: Data Science represents a major structural gap because the program classification is too new to have established tracking. However, for the remaining programs, the data is missing due to federal privacy suppression thresholds: 88% of these missing rows have 10 or fewer graduates, with a median cohort size of just 2.5 students, triggering mandatory redaction by the Census Bureau."

```python
# ==============================================================================
# CORRECTED STEP 1: Filter using your verified PSEO status indicators
# ==============================================================================
# Filtering for rows that were completely unmatched by PSEO 
missing_y1 = merged_cleaned[merged_cleaned['has_pseo_match'] == False].copy()

# Note: If you want to check all rows lacking hard earnings numbers (including those 
# suppressed due to small sizes), you can alternative look at:
# missing_y1 = merged_cleaned[merged_cleaned['y1_earnings_program_status'] == 'not provided by PSEO'].copy()
# ==============================================================================

print(f"Total Rows Missing Year-1 Earnings via has_pseo_match == False: {len(missing_y1)}\n")

# --- DIAGNOSTIC 1: Operational Status (New vs. Discontinued) ---
print("=== 1. Operational Status Breakdown (CURROPER) ===")
print(missing_y1['CURROPER'].value_counts(dropna=False))
print("-" * 50)

# --- DIAGNOSTIC 2: Cohort Graduate Counts (Privacy Suppression) ---
print("=== 2. Cohort Graduate Counts (IPEDSCOUNT1 & IPEDSCOUNT2) ===")
print(missing_y1[['IPEDSCOUNT1', 'IPEDSCOUNT2']].describe())

# Count tiny cohorts where Census Bureau privacy suppression triggers
zero_cohort = missing_y1[missing_y1['IPEDSCOUNT1'] == 0].shape[0]
small_cohort = missing_y1[(missing_y1['IPEDSCOUNT1'] > 0) & (missing_y1['IPEDSCOUNT1'] < 11)].shape[0]

print(f"\nRows with exactly 0 Year-1 Graduates: {zero_cohort}")
print(f"Rows with 1-10 Year-1 Graduates (Likely suppressed for privacy): {small_cohort}")
print("-" * 50)

# --- DIAGNOSTIC 3: Deep Dive into Rows PSEO Doesn't Cover at All ---
missing_both = missing_y1[missing_y1['y1_earnings_program_status'] == 'not provided by PSEO']

print(f"=== 3. Deep Dive: Rows Not Provided by PSEO ===")
print("Top 10 Program Categories (CIPDESC) not provided by PSEO:")
print(missing_both['CIPDESC'].value_counts().head(10))
print("-" * 50)

# --- DIAGNOSTIC 4: Benchmark Availability ---
print("=== 4. Do they at least have National Benchmarks? ===")
print("Has National Benchmark Y1 Status:")
print(missing_y1['y1_earnings_natl_benchmark_level'].value_counts(dropna=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# 1. Filter out the missing rows exactly as we did in your script
missing_y1 = merged_cleaned[merged_cleaned['has_pseo_match'] == False].copy()

# 2. Define clear bins highlighting the privacy cutoff threshold (< 11)
# Bins: 0, 1-5, 6-10, and 11+
bins = [-1, 0, 5, 10, max(missing_y1['IPEDSCOUNT1'].max(), 11)]
labels = ['0 Graduates\n(New/Untracked)', '1–5 Graduates\n(Suppressed)', '6–10 Graduates\n(Suppressed)', '11+ Graduates\n(Niche/Unmapped)']

# 3. Categorize the rows
missing_y1['Cohort Group'] = pd.cut(missing_y1['IPEDSCOUNT1'], bins=bins, labels=labels)
plot_data = missing_y1['Cohort Group'].value_counts().reindex(labels).reset_index()
plot_data.columns = ['Cohort Size', 'Number of Programs']

# 4. Generate a clean, academic-style bar plot
plt.figure(figsize=(9, 5.5))
sns.set_theme(style="whitegrid")

# Use a distinct color palette to isolate the privacy suppression zone
colors = ['#4A5568', '#E53E3E', '#F6AD55', '#3182CE']
ax = sns.barplot(x='Cohort Size', y='Number of Programs', data=plot_data, palette=colors)

# 5. Add value labels on top of each bar for easy reading
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='baseline', 
                fontsize=11, color='black', 
                xytext=(0, 5), textcoords='offset points')

# 6. Title and Labels styling
plt.title('Distribution of Graduate Counts for Missing Earnings Rows\n(Highlighting Privacy Suppression Thresholds)', 
          fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Graduate Cohort Size (IPEDSCOUNT1)', fontsize=12, labelpad=10)
plt.ylabel('Number of Active Programs', fontsize=12, labelpad=10)

# Clean up layout boundaries
plt.tight_layout()

# Save the visualization directly to your homework or project folder
#plt.savefig('../../../data/processed/IPEDS/missing_earnings_suppression_plot.png', dpi=300)
plt.show()
```
