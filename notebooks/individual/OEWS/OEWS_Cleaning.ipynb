# OEWS - Cleaning Data

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
file_path = Path("../../data/raw/OWES/OWES_2025.csv")
df = pd.read_csv(file_path)

np.shape(df)
```

```python
print(df.columns)
```

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(r"\(\d+\)", "", regex=True).str.replace("%", "percent", regex=False).str.replace(r"[^a-z0-9]+", "_", regex=True).str.strip("_"))

df.columns.tolist()
```

```python
df.dtypes
```

```python
df= df[["occ_code", "naics"]].drop_duplicates()
```

```python
df.columns
```

```python
df["year"] = 2025
```

```python
df.columns
```

```python
df.head(20)
```

```python
output_path = Path("../../data/processed/OEWS/OEWS_2025_cleaned.csv")

output_path.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(output_path, index=False)

print(f"Saved cleaned dataset to: {output_path}")
print("Final shape:", df.shape)
```
