BLS - Cleaning Data

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
file_path = Path("../data/raw/BLS/BLS_2025(OES Sheet).csv")
df = pd.read_csv(file_path, header=2)

np.shape(df)
```

```python
print(df.columns)
```

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(r"\(\d+\)", "", regex=True).str.replace("%", "percent", regex=False).str.replace(r"[^a-z0-9]+", "_", regex=True).str.strip("_"))

df.columns.tolist()
```

```python
numeric_columns = [
    "employment",
    "employment_percent_relative_standard_error",
    "hourly_mean_wage",
    "annual_mean_wage",
    "wage_percent_relative_standard_error",
    "hourly_10th_percentile_wage",
    "hourly_25th_percentile_wage",
    "hourly_median_wage",
    "hourly_75th_percentile_wage",
    "hourly_90th_percentile_wage",
    "annual_10th_percentile_wage",
    "annual_25th_percentile_wage",
    "annual_median_wage",
    "annual_75th_percentile_wage",
    "annual_90th_percentile_wage"
]
```

```python
for col in numeric_columns:
    df[col] = (df[col].astype(str).str.replace("$", "", regex=False).str.replace(",", "", regex=False).str.strip())

    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
missing = pd.DataFrame({"missing_count": df.isna().sum()})

missing.sort_values("missing_count", ascending=False)
```

```python
missing_rows = df[df.isna().any(axis=1)]

missing_rows
```

```python
df = df.iloc[:444].copy()

df.shape
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df[numeric_columns].describe().T
```

```python
skewness = df[numeric_columns].skew().sort_values(ascending=False)

skewness
```

```python
outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_summary.append({"feature": col, "outliers": len(outliers), "percent": round(len(outliers) / len(df) * 100, 2)})

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
hourly_columns = ["hourly_mean_wage", "hourly_10th_percentile_wage", "hourly_25th_percentile_wage", "hourly_median_wage", "hourly_75th_percentile_wage", "hourly_90th_percentile_wage"]

df = df.drop(columns=hourly_columns)

print(df.shape)
df.columns.tolist()
```

```python
df["log_employment"] = np.log1p(df["employment"])

print("Original employment skewness:", df["employment"].skew())
print("Log employment skewness:", df["log_employment"].skew())
```

```python
wage_columns = [
    "annual_mean_wage",
    "annual_10th_percentile_wage",
    "annual_25th_percentile_wage",
    "annual_median_wage",
    "annual_75th_percentile_wage",
    "annual_90th_percentile_wage"
]

for col in wage_columns:
    lower = df[col].quantile(0.01)
    upper = df[col].quantile(0.99)

    df[f"{col}_winsorized"] = df[col].clip(
        lower=lower,
        upper=upper
    )
```

```python
df["wage_spread"] = (
    df["annual_90th_percentile_wage"]
    - df["annual_10th_percentile_wage"]
)

df["wage_progression_ratio"] = (
    df["annual_90th_percentile_wage"]
    / df["annual_10th_percentile_wage"]
)
```

```python
import sklearn
from sklearn.preprocessing import StandardScaler

features_to_normalize = [
    "log_employment",
    "annual_mean_wage_winsorized",
    "annual_10th_percentile_wage_winsorized",
    "annual_25th_percentile_wage_winsorized",
    "annual_median_wage_winsorized",
    "annual_75th_percentile_wage_winsorized",
    "annual_90th_percentile_wage_winsorized",
    "wage_spread",
    "wage_progression_ratio"
]

scaler = StandardScaler()

df[[f"{col}_normalized" for col in features_to_normalize]] = (
    scaler.fit_transform(df[features_to_normalize])
)
```
